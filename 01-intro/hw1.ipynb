```python
asia = file[file['origin']=='Asia'][['vehicle_weight', 'model_year']]
X = np.array(asia.head(7))
XTX = X.T@X
inv_XTX = np.linalg.inv(XTX)
y = [1100, 1300, 800, 900, 1000, 1100, 1200]
w = inv_XTX@X.T@y
```
